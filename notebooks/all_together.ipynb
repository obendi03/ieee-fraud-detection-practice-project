```python
import os
import sys
from dotenv import load_dotenv
from pathlib import Path 

notebook_dir = Path.cwd()
ENV_PATH = notebook_dir.parent / '.env'
load_dotenv(dotenv_path=ENV_PATH,override=True)
PATH_TO_DATA = "../data"
os.makedirs(PATH_TO_DATA,exist_ok=True)
```

```python
from kaggle.api.kaggle_api_extended import KaggleApi

api = KaggleApi()
api.authenticate()

competition = "ieee-fraud-detection"

api.competition_download_files(
    competition,
    path=PATH_TO_DATA
)
```

```python
import zipfile
PATH_TO_ZIP = PATH_TO_DATA / Path(os.listdir(PATH_TO_DATA)[0])
with zipfile.ZipFile(PATH_TO_ZIP, 'r') as zip_ref:
    zip_ref.extractall(PATH_TO_DATA)
```

```python
import pandas as pd

PATH_TO_TRAIN_TRANSACTION = PATH_TO_DATA / Path("train_transaction.csv")
train_transaction_df = pd.read_csv(PATH_TO_TRAIN_TRANSACTION)
train_transaction_df.head(5)
```

```python
PATH_TO_TRAIN_IDENTITY = PATH_TO_DATA / Path("train_identity.csv")
train_identity_df = pd.read_csv(PATH_TO_TRAIN_IDENTITY)
train_identity_df.head(5)
```
